```python
age_list = [24, 25, 26, 27, 28]

name_list = ["Kavya", "Anu", "Shalini", "Kiran", "Manu"]

print(age_list)
print(name_list)
```

```python
name_list.append("Yazhini")
print("After append:", name_list)
```

```python
age_list.insert(2, 30)
print("After insert:", age_list)
```

```python
name_list.remove("Yazhini")
print("After remove:", name_list)
```

```python
age_list.pop()
print("After pop:", age_list)
```

```python
age_list.extend([29, 30, 26])
print("After extend:", age_list)
```

```python
age_list.sort(reverse=True)
print("Descending order:", age_list)
```

```python
print("Maximum age:", max(age_list))
```

```python
print("Minimum age:", min(age_list))
```

```python
print("Sum of ages:", sum(age_list))
```

```python
print("First element:", name_list[0])
```

```python
print("Index 2 to 4:", name_list[2:5])
```

```python
print("Reverse order:", name_list[::-1])
```

```python
student_marks = {
    "Kavya": 85,
    "Anu": 78,
    "Shalini": 90,
    "Kiran": 88,
    "Manu": 75
}

print(student_marks)
```

```python
print("Kavya's mark:", student_marks["Kavya"])
```

```python
student_marks["Anu"] = 82
print("After updating Anu:", student_marks)
```

```python
print("Keys:", student_marks.keys())
print("Values:", student_marks.values())
print("Items:", student_marks.items())
```

```python
my_set = {'a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'}

print(my_set)
```

```python
#A set stores only unique values. Duplicate values such as 'a' and 'i' are removed automatically.
#The order of elements in a set is not fixed.
```

```python
my_set[4] = 's'
```

```python
#A set is unordered and does not support indexing.
#Therefore, an element cannot be accessed or changed using an index such as my_set[4].
```

```python
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}

print("Union:", set1.union(set2))
print("Intersection:", set1.intersection(set2))
```
