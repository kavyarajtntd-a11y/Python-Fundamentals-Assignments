```python
import numpy as np
import pandas as pd
```

```python
# Part 1: NumPy Array Operations
```

```python
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print("Week 1 Temperatures:")
print(temperatures_w1)
```

```python
print("Shape:", temperatures_w1.shape)
print("Data Type:", temperatures_w1.dtype)
print("Number of Elements:", temperatures_w1.size)
```

```python
temperatures_f = (temperatures_w1 * 9/5) + 32

print("Temperatures in Fahrenheit:")
print(temperatures_f)
```

```python
print("Maximum Temperature:", np.max(temperatures_w1))
print("Minimum Temperature:", np.min(temperatures_w1))
print("Mean Temperature:", np.mean(temperatures_w1))
```

```python
first_three_days = temperatures_w1[:3]

print("First Three Days:", first_three_days)
```

```python
weekend = temperatures_w1[-2:]

print("Weekend Temperatures:", weekend)
```

```python
middle_three_days = temperatures_w1[2:5]

print("Middle Three Days:", middle_three_days)
```

```python
temperatures = np.array([[22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],[19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]])

print("Temperatures for Two Weeks:")
print(temperatures)
```

```python
print("Shape:", temperatures.shape)
print("Data Type:", temperatures.dtype)
print("Total Number of Elements:", temperatures.size)
```

```python
week1 = temperatures[0, :]

print("Week 1 Temperatures:", week1)
```

```python
week2 = temperatures[1, :]

print("Week 2 Temperatures:", week2)
```

```python
weekends = temperatures[:, -2:]

print("Weekend Temperatures for Both Weeks:")
print(weekends)
```

```python
#Part 2: Pandas Series
```

```python
marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

print("Student Marks:")
print(marks)
```

```python
first_rank = marks.iloc[0]

print("First Rank Marks:", first_rank)
```

```python
top_three = marks.loc['Rank1':'Rank3']

print("Top 3 Students:")
print(top_three)
```

```python
third_rank = marks.iloc[2]

print("Third Rank Marks:", third_rank)
```

```python
high_marks = marks[marks > 90]

print("Students with Marks Greater Than 90:")
print(high_marks)
```

```python
marks.loc['Rank1'] = 100

print("Updated Student Marks:")
print(marks)
```

```python
marks = marks.drop('Rank5')

print("Marks After Removing Rank5:")
print(marks)
```

```python
cgpa = marks / 10

print("Student CGPA:")
print(cgpa)
```

```python
#Part 3: Pandas DataFrame
```

```python
transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],

    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics',
        'Furniture', 'Clothing', 'Electronics',
        'Furniture', 'Clothing', 'Furniture', 'Electronics'
    ],

    'Region': [
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'
    ],

    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})
transactions
```

```python
print("First 5 Rows:")
display(transactions.head())
```

```python
print("Last 5 Rows:")
display(transactions.tail())
```

```python
print("Shape:", transactions.shape)
```

```python
print("Column Names:", transactions.columns)
```

```python
print("Data Types:")
print(transactions.dtypes)
```

```python
print("DataFrame Information:")
transactions.info()
```

```python
selected_columns = transactions[['ProductCategory', 'Amount']]

print("Selected Columns:")
display(selected_columns)
```

```python
last_three_columns = transactions.iloc[:, -3:]

print("Last Three Columns:")
display(last_three_columns)
```

```python
filtered_transactions = transactions[
    (transactions['Region'] == 'North') &
    (transactions['Amount'] > 200)
]

print("Filtered Transactions:")
display(filtered_transactions)
```

```python
category_counts = transactions['ProductCategory'].value_counts()

print("Product Category Counts:")
print(category_counts)
```

```python
unique_regions = transactions['Region'].unique()

print("Unique Regions:")
print(unique_regions)
```

```python
region_mean = transactions.groupby('Region')['Amount'].mean()

print("Average Transaction Amount by Region:")
print(region_mean)
```

```python
transactions.loc[
    transactions['TransactionID'] == 102, 'Amount'
] = 165

print("Updated Transaction:")
display(transactions[transactions['TransactionID'] == 102])
```

```python
transactions['Discount'] = transactions['Amount'] * 0.10

print("Transactions with Discount:")
display(transactions)
```

```python
transactions = transactions.drop(
    transactions[transactions['TransactionID'] == 109].index
)

print("Transactions After Removing TransactionID 109:")
display(transactions)
```

```python
transactions = transactions.drop(columns=['Discount'])

print("Transactions After Removing Discount Column:")
display(transactions)
```

```python
print("Final DataFrame Shape:", transactions.shape)
print("Final Column Names:", transactions.columns.tolist())
```
